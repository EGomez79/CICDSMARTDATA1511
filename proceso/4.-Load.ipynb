```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import functions as F
```

```python
dbutils.widgets.text("catalogo", "catalog_au")
dbutils.widgets.text("esquema_source", "silver")
dbutils.widgets.text("esquema_sink", "golden")
```

```python
catalogo = dbutils.widgets.get("catalogo")
esquema_source = dbutils.widgets.get("esquema_source")
esquema_sink = dbutils.widgets.get("esquema_sink")
```

```python
df_movie_transformed = spark.table(f"{catalogo}.{esquema_source}.movie_transformed")
```

```python
##df_transformed = df_movie_transformed.groupBy(col("yearReleaseDate")).agg(
##                                                count(col("movieId")).alias("conteo"),
##                                               max(col("title")).alias("max_title"),     
##                                                max(col("budget")).alias("max_budget"),
##                                                min(col("budget")).alias("min_budget"),
##                                               max(col("yearReleaseDate")).alias("max_yearReleaseDate"), 
##                                                max(col("durationTime")).alias("max_durationTime"),
##                                                max(col("characterName")).alias("max_characterName"),
##                                               ).orderBy(col("yearReleaseDate").desc())

df_transformed = df_movie_transformed.groupBy(col("yearReleaseDate")).agg(
                                                count(col("movieId")).alias("conteo"),
                                                max(col("title")).alias("max_title"),     
                                                max(col("budget")).alias("max_budget"),
                                                max(col("yearReleaseDate")).alias("max_yearReleaseDate"),
                                                max(col("genderId")).alias("max_genderId"),
                                                max(col("durationTime")).alias("max_durationTime"),
                                                max(col("characterName")).alias("max_characterName"),
                                                ).orderBy(col("yearReleaseDate").desc())
```

```python
df_transformed.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(f"{catalogo}.{esquema_sink}.movie_transformed")
```
